```python
import pandas as pd
import folium
from folium.plugins import HeatMap

df = pd.read_csv('thai_accidental_dataset.csv')

# ใส่ fatalities เป็นค่าน้ำหนักที่ 3 ของแต่ละจุด แทนที่จะนับเป็น 1 เท่ากันหมด
heat_data_severity = df[['latitude', 'longitude', 'fatalities']].values.tolist()

m2 = folium.Map(location=[13.7563, 100.5018], zoom_start=6)
HeatMap(heat_data_severity, radius=8, blur=10, max_zoom=13).add_to(m2)
m2.save('heatmap_severity.html')
```

```python
from thefuzz import fuzz
import itertools

causes = sorted(df['cause_clean'].unique())

similar_pairs = []
for a, b in itertools.combinations(causes, 2):
    score = fuzz.ratio(a, b)
    if score >= 85:
        similar_pairs.append((a, b, score))

for pair in sorted(similar_pairs, key=lambda x: -x[2]):
    print(pair)
```
